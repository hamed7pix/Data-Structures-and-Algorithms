**Backtracking for N-Queens- Algorithm Skeleton (Python-style)**

```python
def is_safe(board, row, col):
    for r in range(row):
        c = board[r]
        if c == col or abs(c - col) == abs(r - row):
            return False
    return True

def solve(row):
    if row == N:                 # all queens placed → solution found
        record the solution
        return

    for col in range(N):         # try every column
        if is_safe(board, row, col):
            board[row] = col     # place queen
            solve(row + 1)       # recurse to next row
            board[row] = -1      # backtrack (undo)
```

**Backtracking for N-Queens**

This 1D array implementation is highly space-efficient, using $O(N)$ space instead of $O(N^2)$ for the board.

```python
def solve_n_queens(N):
    solutions = []
    # 1D array: index represents the row, value represents the column
    board = [-1] * N

    def is_safe(board, row, col):
        for r in range(row):
            c = board[r]
            # Check for column conflict or diagonal conflict
            if c == col or abs(c - col) == abs(r - row):
                return False
        return True

    def solve(row):
        if row == N:                 # all queens placed → solution found
            # Record the solution: convert the 1D array into a 2D string grid
            formatted_solution = []
            for r in range(N):
                row_string = ['.'] * N
                row_string[board[r]] = 'Q'
                formatted_solution.append("".join(row_string))

            solutions.append(formatted_solution)
            return

        for col in range(N):         # try every column
            if is_safe(board, row, col):
                board[row] = col     # place queen
                solve(row + 1)       # recurse to next row
                board[row] = -1      # backtrack (undo)

    # Start the backtracking process from row 0
    solve(0)
    return solutions
```

```python
# --- Execution and Output ---
N = 4
results = solve_n_queens(N)

print(f"Total solutions for {N}-Queens: {len(results)}\n")
for i, solution in enumerate(results):
    print(f"Solution {i + 1}:")
    for row in solution:
        print(row)
    print()
```
