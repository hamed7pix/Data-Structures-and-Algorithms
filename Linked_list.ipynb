# Linked List Operations in Python

This notebook implements a **Singly Linked List** with the following operations:

1. **Insertion**
   - At head (beginning)
   - At tail (end)
   - At a given position

2. **Deletion**
   - From beginning (head)
   - From end (tail)
   - From a given position
   - By value

3. **Search / Lookup**
   - Search by value
   - Search by position

4. **Traversal**
   - Print all nodes
   - Count nodes / get length

## 1. Node and LinkedList Classes

```python
class Node:
    """A single node in the linked list."""
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    """Singly Linked List implementation."""

    def __init__(self):
        self.head = None

    # -------------------------------------------------
    # 4. TRAVERSAL
    # -------------------------------------------------
    def traverse(self):
        """Visit every node and print the list."""
        if self.head is None:
            print("List is empty")
            return

        current = self.head
        elements = []
        while current:
            elements.append(str(current.data))
            current = current.next
        print(" -> ".join(elements) + " -> None")

    def length(self):
        """Return the number of nodes in the list."""
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def is_empty(self):
        """Check if the list is empty."""
        return self.head is None

    # -------------------------------------------------
    # 1. INSERTION
    # -------------------------------------------------
    def insert_at_head(self, data):
        """Insert a new node at the beginning (head)."""
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node
        print(f"Inserted {data} at head")

    def insert_at_tail(self, data):
        """Insert a new node at the end (tail)."""
        new_node = Node(data)

        if self.head is None:
            self.head = new_node
            print(f"Inserted {data} at tail (list was empty)")
            return

        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
        print(f"Inserted {data} at tail")

    def insert_at_position(self, data, position):
        """
        Insert a new node at a given position (0-based index).
        position = 0 means insert at head.
        """
        if position < 0:
            print("Invalid position (negative)")
            return

        if position == 0:
            self.insert_at_head(data)
            return

        new_node = Node(data)
        current = self.head
        count = 0

        # Traverse to the node just before the desired position
        while current and count < position - 1:
            current = current.next
            count += 1

        if current is None:
            print(f"Position {position} is out of bounds. List length is {self.length()}")
            return

        new_node.next = current.next
        current.next = new_node
        print(f"Inserted {data} at position {position}")

    # -------------------------------------------------
    # 2. DELETION
    # -------------------------------------------------
    def delete_from_beginning(self):
        """Delete the node at the head."""
        if self.head is None:
            print("List is empty. Nothing to delete.")
            return

        deleted_data = self.head.data
        self.head = self.head.next
        print(f"Deleted {deleted_data} from beginning")

    def delete_from_end(self):
        """Delete the node at the tail."""
        if self.head is None:
            print("List is empty. Nothing to delete.")
            return

        # Only one node
        if self.head.next is None:
            deleted_data = self.head.data
            self.head = None
            print(f"Deleted {deleted_data} from end (only node)")
            return

        # Find the second last node
        current = self.head
        while current.next.next:
            current = current.next

        deleted_data = current.next.data
        current.next = None
        print(f"Deleted {deleted_data} from end")

    def delete_at_position(self, position):
        """Delete the node at a given position (0-based index)."""
        if self.head is None:
            print("List is empty. Nothing to delete.")
            return

        if position < 0:
            print("Invalid position (negative)")
            return

        # Delete head
        if position == 0:
            self.delete_from_beginning()
            return

        current = self.head
        count = 0

        # Traverse to the node just before the one to delete
        while current and count < position - 1:
            current = current.next
            count += 1

        if current is None or current.next is None:
            print(f"Position {position} is out of bounds. List length is {self.length()}")
            return

        deleted_data = current.next.data
        current.next = current.next.next
        print(f"Deleted {deleted_data} from position {position}")

    def delete_by_value(self, value):
        """Delete the first occurrence of a node with the given value."""
        if self.head is None:
            print("List is empty. Nothing to delete.")
            return

        # If head needs to be deleted
        if self.head.data == value:
            self.head = self.head.next
            print(f"Deleted value {value} (was at head)")
            return

        current = self.head
        while current.next and current.next.data != value:
            current = current.next

        if current.next is None:
            print(f"Value {value} not found in the list")
            return

        current.next = current.next.next
        print(f"Deleted value {value}")

    # -------------------------------------------------
    # 3. SEARCH / LOOKUP
    # -------------------------------------------------
    def search_by_value(self, value):
        """
        Search for a value in the list.
        Returns the position (0-based) if found, else -1.
        """
        current = self.head
        position = 0

        while current:
            if current.data == value:
                print(f"Value {value} found at position {position}")
                return position
            current = current.next
            position += 1

        print(f"Value {value} not found")
        return -1

    def search_by_position(self, position):
        """
        Return the data at a given position (0-based index).
        Returns None if position is invalid.
        """
        if position < 0:
            print("Invalid position (negative)")
            return None

        current = self.head
        count = 0

        while current:
            if count == position:
                print(f"Data at position {position} is {current.data}")
                return current.data
            current = current.next
            count += 1

        print(f"Position {position} is out of bounds. List length is {self.length()}")
        return None
```

## 2. Demo / Testing the Operations

```python
# Create an empty linked list
ll = LinkedList()

print("=== Initial state ===")
ll.traverse()
print(f"Length: {ll.length()}")
print(f"Is empty? {ll.is_empty()}")
```

### Insertion Examples

```python
print("=== Insertion at Head ===")
ll.insert_at_head(30)
ll.insert_at_head(20)
ll.insert_at_head(10)
ll.traverse()

print("\n=== Insertion at Tail ===")
ll.insert_at_tail(40)
ll.insert_at_tail(50)
ll.traverse()

print("\n=== Insertion at Position ===")
ll.insert_at_position(25, 2)   # Insert 25 at index 2
ll.insert_at_position(5, 0)    # Insert 5 at head (position 0)
ll.insert_at_position(60, 7)   # Insert at the end (current length will be 7 after previous inserts)
ll.traverse()
print(f"Current length: {ll.length()}")
```

### Search Examples

```python
print("=== Search by Value ===")
ll.search_by_value(25)
ll.search_by_value(100)   # not present

print("\n=== Search by Position ===")
ll.search_by_position(0)
ll.search_by_position(3)
ll.search_by_position(20)  # out of bounds
```

### Deletion Examples

```python
print("=== Current List ===")
ll.traverse()

print("\n=== Delete from Beginning ===")
ll.delete_from_beginning()
ll.traverse()

print("\n=== Delete from End ===")
ll.delete_from_end()
ll.traverse()

print("\n=== Delete at Position ===")
ll.delete_at_position(2)
ll.traverse()

print("\n=== Delete by Value ===")
ll.delete_by_value(40)
ll.traverse()

print("\n=== Final state ===")
print(f"Length: {ll.length()}")
ll.traverse()
```

## Summary of Methods

| Operation              | Method                      | Description                          |
|------------------------|-----------------------------|--------------------------------------|
| Insert at head         | `insert_at_head(data)`      | Add node at the beginning            |
| Insert at tail         | `insert_at_tail(data)`      | Add node at the end                  |
| Insert at position     | `insert_at_position(data, pos)` | Add node at given index          |
| Delete from beginning  | `delete_from_beginning()`   | Remove head node                     |
| Delete from end        | `delete_from_end()`         | Remove tail node                     |
| Delete at position     | `delete_at_position(pos)`   | Remove node at given index           |
| Delete by value        | `delete_by_value(value)`    | Remove first occurrence of value     |
| Search by value        | `search_by_value(value)`    | Return position or -1                |
| Search by position     | `search_by_position(pos)`   | Return data at index                 |
| Traverse / Print       | `traverse()`                | Print the entire list                |
| Length                 | `length()`                  | Return number of nodes               |
| Is empty               | `is_empty()`                | Check if list has no nodes           |
