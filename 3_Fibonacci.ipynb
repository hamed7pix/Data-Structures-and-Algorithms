This is the standard, recursive implementation. When you run this, Python's internal call stack automatically handles all the PUSH and POP state management that was manually explicitly tracked in the previous examples. Because it recalculates the same branches multiple times (like fib(2)), it has an exponential time complexity of $O(2^n)$.

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
print(fibonacci(4))
```

**You want to see the actual contents of the stack array as it grows (PUSH) and shrinks (POP), demonstrating the Last-In-First-Out (LIFO) memory behavior, rather than looking at the shape of the tree.**

```python
def fibonacci_stack_state(n):
    # The explicit stack holds the frames
    stack = [{'n': n, 'stage': 0, 'left': None, 'right': None}]
    return_value = None

    def print_stack(action):
        # Format the current items in the stack as a list
        stack_str = "[" + ", ".join([f"fib({f['n']})" for f in stack]) + "]"
        if not stack:
            stack_str = "[]"
        print(f"{action:<16} | Stack: {stack_str}")

    print("--- Starting Execution ---")
    print_stack("INIT")

    while stack:
        frame = stack[-1] # Peek at the top of the stack
        curr_n = frame['n']

        # Stage 0: Initial evaluation
        if frame['stage'] == 0:
            if curr_n <= 1:
                return_value = curr_n
                stack.pop() # Base case resolves, remove from stack
                print_stack(f"POP (return {curr_n})")
            else:
                frame['stage'] = 1
                stack.append({'n': curr_n - 1, 'stage': 0, 'left': None, 'right': None})
                print_stack(f"PUSH fib({curr_n - 1})")

        # Stage 1: Left branch finished, pushing right branch
        elif frame['stage'] == 1:
            frame['left'] = return_value
            frame['stage'] = 2
            stack.append({'n': curr_n - 2, 'stage': 0, 'left': None, 'right': None})
            print_stack(f"PUSH fib({curr_n - 2})")

        # Stage 2: Both branches finished, calculate and pop
        elif frame['stage'] == 2:
            frame['right'] = return_value
            result = frame['left'] + frame['right']
            return_value = result
            stack.pop() # Frame complete, remove from stack
            print_stack(f"POP (return {result})")

    print("--- Execution Complete ---")
    return return_value
```

```python
fibonacci_stack_state(3)
```

**A clean implementation that prints a pure visual tree structure of the Fibonacci calls.**

```python
def print_fibonacci_tree(n, prefix="", is_left=True, is_root=True):
    # Print the current node
    if is_root:
        print(f"fib({n})")
    else:
        # Use a T-branch for the left child, and an L-branch for the right child
        branch = "├── " if is_left else "└── "
        print(f"{prefix}{branch}fib({n})")

    # Base case: stop branching
    if n <= 1:
        return

    # Calculate the prefix for the next level down
    if is_root:
        new_prefix = ""
    else:
        # Left children drop a vertical line down for their siblings, right children leave a blank space
        new_prefix = prefix + ("│   " if is_left else "    ")

    # Recursive calls for left (n-1) and right (n-2) branches
    print_fibonacci_tree(n - 1, new_prefix, is_left=True, is_root=False)
    print_fibonacci_tree(n - 2, new_prefix, is_left=False, is_root=False)
```

```python
# Test the function
print_fibonacci_tree(4)
```
