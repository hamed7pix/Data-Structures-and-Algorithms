```python
def constrained_maximum_subarray_v1(arr, L, R, lo, hi):
    if not arr:
        raise ValueError("arr must not be empty")

    if not (1 <= L <= R <= len(arr)):
        raise ValueError("Require 1 <= L <= R <= len(arr)")

    if lo == hi:
        return arr[lo] if L <= 1 <= R else float('-inf')

    mid = (lo + hi) // 2

    left = constrained_maximum_subarray_v1(arr, L, R, lo, mid)
    right = constrained_maximum_subarray_v1(arr, L, R, mid + 1, hi)
    cross = crossing_v1(arr, L, R, lo, mid, hi)

    return max(
        left,
        right,
        cross
    )



def crossing_v1(arr, L, R, lo, mid, hi):
    best = float('-inf')
    for i in range(lo, mid + 1):
        for j in range(mid + 1, hi + 1):
            length = j - i + 1
            if L <= length <= R:
                summation = sum(arr[i : j + 1])
                best = max(best, summation)
    return best
```

        ممکن را بررسی کنیم. برای داده‌های بزرگ هنوز کافی نیست.

```python
from collections import deque


def crossing_v3(P, L, R, lo, mid, hi):
    best = float('-inf')
    candidates = deque()
    next_i = lo

    for j in range(mid + 1, hi + 1):
        max_start = min(mid, j - L + 1)

        while next_i <= max_start:
            while candidates and P[candidates[-1]] >= P[next_i]:
                candidates.pop()

            candidates.append(next_i)
            next_i += 1

        min_start = max(lo, j - R + 1)

        while candidates and candidates[0] < min_start:
            candidates.popleft()

        if candidates:
            i = candidates[0]
            best = max(best, P[j + 1] - P[i])

    return best


def solve_v3(P, L, R, lo, hi):
    if lo == hi:
        return P[lo + 1] - P[lo] if L <= 1 <= R else float('-inf')

    mid = (lo + hi) // 2

    left = solve_v3(P, L, R, lo, mid)
    right = solve_v3(P, L, R, mid + 1, hi)
    cross = crossing_v3(P, L, R, lo, mid, hi)

    return max(left, right, cross)


def constrained_maximum_subarray_v3(arr, L, R):
    if not arr:
        raise ValueError("arr must not be empty")

    if not (1 <= L <= R <= len(arr)):
        raise ValueError("Require 1 <= L <= R <= len(arr)")

    P = [0]
    for x in arr:
        P.append(P[-1] + x)

    return solve_v3(P, L, R, 0, len(arr) - 1)
```

```python
A = [2, -5, 4, -1, 3, -2]
L, R = 2, 4

cms = constrained_maximum_subarray_v3(A, L, R)
print(f'Maximum constrained subarray: {cms}')
```
