# Mastering Two-Way (Doubly) Linked Lists from Scratch

Welcome to this interactive guide on **Doubly Linked Lists** (also known as Two-Way Linked Lists).

Unlike a standard Singly Linked List where each node only knows about the node *after* it, a **Doubly Linked List (DLL)** allows traversal in **both directions** (forward and backward).

### Visual Representation
```text
None <-- [ Prev | Data | Next ] <==> [ Prev | Data | Next ] <==> [ Prev | Data | Next ] --> None
                (Head)                                                  (Tail)
```

### Key Advantages:
- **Bidirectional Traversal**: You can walk from `head` to `tail` or from `tail` to `head`.
- **$O(1)$ Deletion & Insertion at Both Ends**: With a `tail` pointer, adding or removing from the end of the list takes constant time $O(1)$.
- **Easy Node Removal**: Given a reference to a node, you can delete it in $O(1)$ time without needing to search for its predecessor first.

## Step 1: Building the `Node` Class

Every element in a Doubly Linked List is stored inside a `Node`. Each node holds three pieces of information:
1. `data`: The actual value stored in the node.
2. `next`: A pointer/reference to the **next** node in the list (`None` by default).
3. `prev`: A pointer/reference to the **previous** node in the list (`None` by default).

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None  # Reference to the next node
        self.prev = None  # Reference to the previous node

    def __repr__(self):
        return f"Node({self.data})"

# Quick test of a single node
sample_node = Node(42)
print(f"Created: {sample_node}, Prev: {sample_node.prev}, Next: {sample_node.next}")
````

## Step 2: Building the `DoublyLinkedList` Class

Now let's build the complete `DoublyLinkedList` class from scratch. We will maintain references to both `head` (first node) and `tail` (last node) so operations at either end run in $O(1)$ time.

### Operations Included:
1. `append(data)`: Add a node to the **end** of the list — $O(1)$
2. `prepend(data)`: Add a node to the **beginning** of the list — $O(1)$
3. `insert_after(target_data, new_data)`: Insert a new node right after a node containing `target_data` — $O(n)$
4. `delete(target_data)`: Remove the first node containing `target_data` — $O(n)$ to search, $O(1)$ to unlink
5. `display_forward()` & `display_backward()`: Print the list in both directions — $O(n)$

````python
class DoublyLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None
        self.size = 0

    def is_empty(self):
        return self.head is None

    def append(self, data):
        """Insert a new node at the end (tail) of the list."""
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.prev = self.tail
            self.tail.next = new_node
            self.tail = new_node
        self.size += 1

    def prepend(self, data):
        """Insert a new node at the beginning (head) of the list."""
        new_node = Node(data)
        if self.is_empty():
            self.head = new_node
            self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node
        self.size += 1

    def insert_after(self, target_data, new_data):
        """Find the first node with target_data and insert new_data right after it."""
        current = self.head
        while current:
            if current.data == target_data:
                # If target is the tail, just use append
                if current == self.tail:
                    self.append(new_data)
                else:
                    new_node = Node(new_data)
                    new_node.next = current.next
                    new_node.prev = current
                    current.next.prev = new_node
                    current.next = new_node
                    self.size += 1
                return True
            current = current.next
        print(f"Target '{target_data}' not found in list.")
        return False

    def delete(self, target_data):
        """Remove the first node that matches target_data."""
        current = self.head
        while current:
            if current.data == target_data:
                # Case 1: Only one node in the list
                if current == self.head and current == self.tail:
                    self.head = None
                    self.tail = None
                # Case 2: Deleting the head node
                elif current == self.head:
                    self.head = current.next
                    self.head.prev = None
                # Case 3: Deleting the tail node
                elif current == self.tail:
                    self.tail = current.prev
                    self.tail.next = None
                # Case 4: Deleting a middle node
                else:
                    current.prev.next = current.next
                    current.next.prev = current.prev

                self.size -= 1
                return True
            current = current.next
        print(f"Value '{target_data}' not found to delete.")
        return False

    def display_forward(self):
        """Traverse from head to tail."""
        elements = []
        current = self.head
        while current:
            elements.append(str(current.data))
            current = current.next
        print("Forward : None <-> " + " <-> ".join(elements) + " <-> None" if elements else "List is empty")

    def display_backward(self):
        """Traverse from tail to head."""
        elements = []
        current = self.tail
        while current:
            elements.append(str(current.data))
            current = current.prev
        print("Backward: None <-> " + " <-> ".join(elements) + " <-> None" if elements else "List is empty")
````

## Step 3: Testing Basic Operations

Let's test `append`, `prepend`, `insert_after`, `delete`, and bidirectional traversal.

````python
dll = DoublyLinkedList()

print("1. Appending 10, 20, 30:")
dll.append(10)
dll.append(20)
dll.append(30)
dll.display_forward()

print("\n2. Prepending 5:")
dll.prepend(5)
dll.display_forward()

print("\n3. Inserting 15 after 10:")
dll.insert_after(10, 15)
dll.display_forward()
dll.display_backward()

print("\n4. Deleting 20 (middle node) and 5 (head node):")
dll.delete(20)
dll.delete(5)
dll.display_forward()
dll.display_backward()
print(f"Current size: {dll.size}")
````

## Step 4: Real-World Example — Browser History (Back & Forward Navigation)

One of the most classic real-world applications of a **Two-Way Linked List** is a web browser's **Back and Forward navigation** (or an Undo/Redo system in a text editor).

### How it works:
- Each webpage you visit is a `Node`.
- Clicking **Back** moves the `current_page` pointer to `current_page.prev`.
- Clicking **Forward** moves the `current_page` pointer to `current_page.next`.
- Visiting a **new URL** while in the middle of history clears the forward history and links the new page right after `current_page`.

````python
class BrowserHistory:
    def __init__(self, homepage: str):
        self.current = Node(homepage)
        print(f"Opened browser at: {self.current.data}")

    def visit(self, url: str):
        """Visit a new URL from the current page (clears any forward history)."""
        new_page = Node(url)
        self.current.next = new_page
        new_page.prev = self.current
        self.current = new_page
        print(f"Visited: {self.current.data}")

    def back(self, steps: int = 1):
        """Move backward in history by `steps`."""
        while steps > 0 and self.current.prev is not None:
            self.current = self.current.prev
            steps -= 1
        print(f"<- Back to: {self.current.data}")
        return self.current.data

    def forward(self, steps: int = 1):
        """Move forward in history by `steps`."""
        while steps > 0 and self.current.next is not None:
            self.current = self.current.next
            steps -= 1
        print(f"-> Forward to: {self.current.data}")
        return self.current.data

    def show_history_state(self):
        """Visualize the entire history chain and highlight the active page."""
        # Find the very first page
        temp = self.current
        while temp.prev:
            temp = temp.prev

        chain = []
        while temp:
            if temp == self.current:
                chain.append(f"[{temp.data} (CURRENT)]")
            else:
                chain.append(temp.data)
            temp = temp.next
        print("History Chain: " + " <==> ".join(chain))

# --- Simulate a user browsing session ---
browser = BrowserHistory("google.com")
browser.visit("wikipedia.org")
browser.visit("python.org")
browser.visit("github.com")
browser.show_history_state()

print("\n--- Navigating Backwards ---")
browser.back(1)  # Back to python.org
browser.back(1)  # Back to wikipedia.org
browser.show_history_state()

print("\n--- Navigating Forward ---")
browser.forward(1)  # Forward to python.org
browser.show_history_state()

print("\n--- Visiting a New Page (Branches History) ---")
browser.visit("stackoverflow.com")  # Replaces github.com in forward history
browser.show_history_state()
````

## Time & Space Complexity Summary

| Operation | Doubly Linked List | Singly Linked List (with Tail) | Dynamic Array (`list`) |
| :--- | :---: | :---: | :---: |
| **Prepend (Insert at Head)** | $O(1)$ | $O(1)$ | $O(n)$ |
| **Append (Insert at Tail)** | $O(1)$ | $O(1)$ | $O(1)$ amortized |
| **Delete at Head** | $O(1)$ | $O(1)$ | $O(n)$ |
| **Delete at Tail** | $O(1)$ | $O(n)$ | $O(1)$ |
| **Delete Given Node Reference** | $O(1)$ | $O(n)$ | $O(n)$ |
| **Search by Value** | $O(n)$ | $O(n)$ | $O(n)$ |
