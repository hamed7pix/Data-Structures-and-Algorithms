# Recursive Algorithms

Recursion is a method where the solution to a problem depends on solutions to smaller instances of the same problem. A recursive function is simply a function that **calls itself**.

Every valid recursive function must have two parts:
1. **The Base Case:** The condition where the recursion stops. Without this, you get an infinite loop (Stack Overflow).
2. **The Recursive Case:** Where the function calls itself with a modified, smaller input.

## 1. The Classic Example: Factorial
The factorial of `n` (written as `n!`) is `n * (n-1) * (n-2) ... * 1`. Notice that `5! = 5 * 4!`. This self-referential property makes it perfect for recursion.

```python
def factorial(n):
    # 1. Base Case: The smallest instance we know the answer to immediately
    if n == 0 or n == 1:
        return 1

    # 2. Recursive Case: Call the function again with a smaller input
    return n * factorial(n - 1)
```

```python
print("5! =", factorial(5))
```

## 2. Visualizing the Call Stack
When a function calls itself, the computer pauses the current function and puts it on a "stack" while it runs the new one. Once the base case is reached, the stack "unwinds" and finishes all the paused functions.

```python
def countdown_and_up(n):
    if n <= 0:
        print("\nBase case reached! Unwinding the stack...\n")
        return

    print(f"Winding down: {n}")
    countdown_and_up(n - 1)  # The recursive call
    print(f"Unwinding up: {n}")

countdown_and_up(3)
```

## 3. Multiple Recursive Calls (The Recursion Tree)
Sometimes a function needs to call itself multiple times. The Fibonacci sequence (`0, 1, 1, 2, 3, 5, 8...`) requires adding the two previous numbers together: `fib(n) = fib(n-1) + fib(n-2)`.

<img src="https://miro.medium.com/1*svQ784qk1hvBE3iz7VGGgQ.jpeg" alt="Fibonacci Recursion Tree" width="600"/>

Notice in the tree above how many times `f(2)` gets recalculated! This branching leads to a massive time complexity of $O(2^n)$.

```python
def fibonacci(n):
    # Base Cases
    if n == 0:
        return 0
    elif n == 1:
        return 1

    # Recursive Case (Branching into two paths)
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
print("The 6th Fibonacci number is:", fibonacci(6))
```
